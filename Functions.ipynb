```python
#Accepts the number from the user and display prime factors of that no.
def print_prime_factors(n):
    i = 2
    print(f"Prime factors of {n} are:", end=" ")
    while i * i <= n:
        while n % i == 0:
            print(i, end=" ")
            n //= i
        i += 1
    if n > 1:
        print(n, end=" ")
    print()

num = int(input("Enter a number: "))
print_prime_factors(num)
```

```python
"""
Write a python program which accepts coefficients of the quadratic equation and generates real roots(with same or different values).
If (b^2-4ac)<0, use of math.sqrt function generate an execption an Value error"math domain error". Show execption handling for ValueError, NameError, and TypeError
"""
import math

def calculate_quadratic_roots():
    print("Quadratic Equation Root Finder (ax² + bx + c = 0)")
    
    try:
        # Accepting coefficients from the user
        # This can raise a ValueError if the input cannot be converted to a float
        a_input = input("Enter coefficient a: ")
        b_input = input("Enter coefficient b: ")
        c_input = input("Enter coefficient c: ")
        
        # Explicit type conversion
        a = float(a_input)
        b = float(b_input)
        c = float(c_input)
        
        # Simulating a TypeError check (e.g., if someone manually passes a wrong type to a function)
        if not all(isinstance(i, (int, float)) for i in [a, b, c]):
            raise TypeError("Coefficients must be numerical values.")
            
        # A quadratic equation cannot have a = 0
        if a == 0:
            print("Coefficient 'a' cannot be zero in a quadratic equation.")
            return

        # Calculating the discriminant
        discriminant = (b ** 2) - (4 * a * c)
        
        # Math.sqrt will raise a ValueError ("math domain error") if discriminant < 0
        root_discriminant = math.sqrt(discriminant)
        
        # Calculating the two real roots
        root1 = (-b + root_discriminant) / (2 * a)
        root2 = (-b - root_discriminant) / (2 * a)
        
        # Displaying the results
        if discriminant == 0:
            print(f"\nThe equation has two equal real roots:\nRoot 1 = Root 2 = {root1:.4f}")
        else:
            print(f"\nThe equation has two distinct real roots:\nRoot 1 = {root1:.4f}\nRoot 2 = {root2:.4f}")

    except ValueError as e:
        # Catches math domain error (negative discriminant) and invalid literal errors (e.g., entering letters)
        if "math domain error" in str(e):
            print("\nError: The equation has complex/imaginary roots. (Discriminant < 0)")
        else:
            print("\nValueError: Please enter valid numerical values for the coefficients.")
            
    except TypeError as e:
        # Catches cases where operations are performed on incompatible data types
        print(f"\nTypeError: An invalid data type operation occurred. Details: {e}")
        
    except NameError as e:
        # Catches cases where a variable name used in the code is not defined
        print(f"\nNameError: A variable name used in the program is not defined. Details: {e}")
        
    except Exception as e:
        # General fallback catch for any other unexpected errors
        print(f"\nAn unexpected error occurred: {e}")

# Run the program
if __name__ == "__main__":
    calculate_quadratic_roots()
```

```python
#Recursion
def print_numbers(n):
    if n==0:  #Base Calculation
        return
    print_numbers(n-1)  #Recursive call
    print(n)
n=int(input("Enter n:"))
print_numbers(n)
```

```python
#Factorial
def factorial(n):
    if n==0 or n==1:
        return 1
        return n*factorial(n-1)
n=int(input("Enter a number:"))
print("Factorial=", factorial(n))
```

```python
#fibonacci
def fibonacci(n):
    if n<=1:
        return n
    return fibonacci(n-1)+fibonacci(n-2)
n=int(input("Enter a number of terms:"))
for i in range(n):
    print(fibonacci(i),end="")
```
